```python
import numpy as np
import matplotlib.pyplot as plt

def elu(x, alpha=1.0):
    x = np.asarray(x, dtype=float)
    return np.where(x > 0, x, alpha * np.expm1(np.minimum(x, 0)))

x = np.linspace(-10, 10, 200)
y = elu(x)

print(elu(np.array([-5, -2, 0, 2, 5])))

plt.figure(figsize=(8, 4))
plt.plot(x, y, label="ELU", linewidth=2)
plt.title("ELU Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output f(x)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def softmax(x):
    x = np.asarray(x, dtype=float)
    exp_x = np.exp(x - np.max(x))
    return exp_x / np.sum(exp_x)

# Class scores (logits)
x = np.array([2.0, 1.0, 0.1])

# Apply Softmax
y = softmax(x)

print("Softmax probabilities:", y)
print("Sum of probabilities:", np.sum(y))

# Visualize class probabilities
classes = ["Cat", "Dog", "Bird"]

plt.figure(figsize=(8, 4))
plt.bar(classes, y)
plt.title("Softmax Activation Function")
plt.xlabel("Classes")
plt.ylabel("Predicted Probability")
plt.ylim(0, 1)
plt.grid(axis="y")
plt.show()
```

# ARTIFICIAL NEURAL NETWORK (ANN)
# Complete Theory, Terminology and Concepts

---

# 1. What Is an Artificial Neural Network (ANN)?

An Artificial Neural Network (ANN) is a machine learning model inspired
by the structure and functioning of biological neural networks.

ANN consists of interconnected computational units called neurons
or perceptrons.

Each neuron receives inputs, performs mathematical calculations,
applies an activation function, and passes its output to other neurons.

ANNs learn relationships between input features and target outputs
by adjusting weights and biases during training.

## Example

Suppose a company wants to predict employee salary based on:

- Years of experience
- Education level
- Technical skills
- Previous salary
- Job location

An ANN learns patterns from historical employee data and predicts
the salary of a new employee.

## Important Characteristics

1. Learns patterns from training data.
2. Supports regression and classification.
3. Models complex nonlinear relationships.
4. Uses weights and biases to learn.
5. Uses activation functions to introduce nonlinearity.
6. Uses loss functions to measure prediction errors.
7. Uses backpropagation to calculate gradients.
8. Uses optimizers to update parameters.
9. Supports multiple hidden layers.
10. Can generalize to previously unseen data.

---

# 2. Biological Neuron vs Artificial Neuron

## Biological Neuron

A biological neuron contains:

1. Dendrites
2. Cell body
3. Axon
4. Synapses

Dendrites receive signals from other neurons.

The cell body processes incoming signals.

The axon transmits signals to other neurons.

Synapses connect neurons and influence signal transmission.

## Artificial Neuron

An artificial neuron contains:

1. Inputs
2. Weights
3. Bias
4. Weighted sum
5. Activation function
6. Output

## Comparison

| Biological Neuron | Artificial Neuron |
|-------------------|-------------------|
| Dendrites | Input features |
| Synapses | Weighted connections |
| Synaptic strength | Weights |
| Cell body | Weighted sum calculation |
| Firing behavior | Activation function |
| Axon output | Neuron output |

Artificial neurons are mathematical abstractions, not exact
simulations of biological neurons.

---

# 3. What Is a Perceptron?

A perceptron is a basic computational unit used in neural networks.

It accepts input features, multiplies them by corresponding weights,
adds a bias, and applies an activation function.

## Mathematical Formula

z = w1*x1 + w2*x2 + ... + wn*xn + b

a = f(z)

Where:

x = Input feature
w = Weight
b = Bias
z = Weighted sum or pre-activation
f = Activation function
a = Activated output

## Example

Inputs:

x1 = 2
x2 = 3

Weights:

w1 = 0.5
w2 = 0.4

Bias:

b = 0.2

Calculation:

z = (2 * 0.5) + (3 * 0.4) + 0.2

z = 1 + 1.2 + 0.2

z = 2.4

If ReLU is used:

a = max(0, 2.4)

a = 2.4

The output of the neuron is 2.4.

## Important Note

The original perceptron uses a threshold activation function.

Modern artificial neural networks typically use differentiable
or almost-everywhere differentiable activation functions.

---

Input features are the independent variables supplied to a neural network.

They represent the information used for prediction.

| E1 | 2 | 24 | 3 | 1 | 30000 |

X1 = Experience

X3 = Skills

Target:

Y = Salary

This dataset has:

5 rows
4 input features
1 target column

The input layer contains 4 input values per observation.

# 5. What Is a Weight?

A weight is a trainable numerical parameter associated with
a connection between neurons.

Weights determine how strongly inputs influence the weighted sum.

## Example

z = w1*x1 + w2*x2 + b

Suppose:

x1 = 5
x2 = 10

w1 = 0.8
w2 = -0.3

b = 1

Then:

z = (5 * 0.8) + (10 * -0.3) + 1

z = 4 - 3 + 1

z = 2

## Interpretation

Positive weight:

Increasing the input increases its contribution to the
weighted sum, assuming other inputs remain constant.

Negative weight:

Increasing the input decreases its contribution.

Zero weight:

The input makes no direct contribution through that connection.

Important:

The magnitude of a weight does not automatically measure
real-world feature importance.

Feature scaling, interactions, and subsequent layers matter.

---

# 6. What Is a Bias?

Bias is a trainable parameter added to the weighted sum.

It allows a neuron to shift its activation independently
of its input contributions.

## Formula

z = wx + b

## Example

Input = 3
Weight = 2
Bias = 5

z = (3 * 2) + 5

z = 11

Without bias:

z = 6

With bias:

z = 11

Bias increases the flexibility of the model.

Usually, each neuron in a Dense layer has one bias parameter.

---

# 7. What Is a Layer?

A layer is a collection of neurons operating at the same
stage of a neural network.

Three major layer categories are:

1. Input layer
2. Hidden layer
3. Output layer

## Architecture

Input Layer
    |
    v
Hidden Layer 1
    |
    v
Hidden Layer 2
    |
    v
Output Layer

Each layer transforms information received from the
previous layer.

---

# 8. Input Layer

The input layer receives feature values from the dataset.

It does not usually perform the trainable weighted-sum
computation in a standard Dense neural network.

## Example

Dataset features:

Age
Experience
Salary
Education
Skills

Number of input features = 5

Input shape for one observation = (5,)

For 100 observations:

Input matrix shape = (100, 5)

100 represents observations.

5 represents input features.

---

# 9. Hidden Layer

A hidden layer is positioned between the input and output layers.

Hidden layers perform learned transformations of input data.

Each hidden neuron generally:

1. Receives inputs.
2. Multiplies inputs by weights.
3. Adds a bias.
4. Applies an activation function.
5. Produces an output.

## Example

Input features = 3

Hidden neurons = 4

Output neurons = 1

Architecture:

Input(3) -> Dense(4) -> Dense(1)

The hidden layer learns intermediate representations.

More hidden layers can increase representational capacity,
but may also increase training complexity and overfitting risk.

---

# 10. Output Layer

The output layer generates the final model prediction.

Its size and activation depend on the problem.

## Regression

Example:

Predict house price.

Output neurons = 1

Activation = Linear

## Binary Classification

Example:

Fraud or Not Fraud

Output neurons = 1

Activation = Sigmoid

## Multiclass Classification

Example:

Cat, Dog, Bird

Output neurons = 3

Activation = Softmax

## Multilabel Classification

Example:

An image may contain a person, car, and bicycle simultaneously.

Output neurons = Number of labels

Activation = Independent Sigmoid for each label

# 11. What Is Network Architecture?

Network architecture describes the organization of a neural network.

It includes:

- Number of input features
- Number of hidden layers
- Number of neurons per layer
- Layer connectivity
- Activation functions
- Output configuration

## Example

Input(5)
   |
Dense(10, ReLU)
   |
Dense(8, ReLU)
   |
Dense(4, ReLU)
   |
Dense(1, Sigmoid)

This architecture contains:

5 input features
3 hidden layers

8 neurons in hidden layer 2

The network is suitable for a binary classification problem.

---

# 12. What Is a Shallow Neural Network?

A shallow neural network contains relatively few trainable layers,
commonly one hidden layer.

## Example

Input(4)
   |
Dense(5, ReLU)
   |
Dense(1, Sigmoid)

Advantages:

- Simpler architecture
- Fewer parameters
- Often faster training
- Easier to experiment with

Limitations:

- May need many neurons for complex patterns
- May not efficiently represent some hierarchical relationships

---

# 13. What Is a Deep Neural Network?

A Deep Neural Network (DNN) contains multiple hidden layers.

## Example

Input(10)
   |
Dense(64, ReLU)
   |
Dense(32, ReLU)
   |
Dense(16, ReLU)
   |
Dense(8, ReLU)
   |
Dense(1, Sigmoid)

A DNN can learn hierarchical representations.

Earlier layers may learn relatively simple transformations.

Later layers combine those representations into more
complex patterns.

Applications:

- Image recognition
- Natural language processing
- Speech recognition
- Fraud detection
- Recommendation systems
- Time-series forecasting

---

# 14. What Is a Fully Connected Layer?

A Fully Connected Layer is also called a Dense layer.

Every neuron in the layer receives an input from every
output unit of the previous layer.

## Example

Input neurons = 3

Hidden neurons = 4

Each of the 3 inputs connects to all 4 hidden neurons.

Number of weights:

3 * 4 = 12

Number of biases:

4

Total parameters:

12 + 4 = 16

## Formula

Parameters = (Inputs * Neurons) + Neurons

Parameters = (Inputs + 1) * Neurons

Hyperparameters are configuration choices that control

They are not normally learned through backpropagation.

## Examples

1. Learning rate
2. Batch size
3. Number of epochs
4. Number of hidden layers

Selected before or during model experimentation.

# 17. What Is an Activation Function?

An activation function transforms a neuron's weighted sum
into its output.

Its primary role in hidden layers is to introduce nonlinearity.

Without nonlinear activation functions, a sequence of
fully connected linear transformations is equivalent to
one linear transformation.

## Formula

z = wx + b

a = f(z)

## Common Activation Functions

| Function | Range | Common Use |
|----------|-------|------------|
| Binary Step | 0 or 1 | Threshold decisions |
| Linear | -Infinity to Infinity | Regression output |
| Sigmoid | 0 to 1 | Binary classification |
| Tanh | -1 to 1 | Traditional RNNs |
| ReLU | 0 to Infinity | Hidden layers |
| Leaky ReLU | -Infinity to Infinity | Hidden layers |
| ELU | -alpha to Infinity | Deep networks |
| Softmax | 0 to 1 | Multiclass classification |
| Swish / SiLU | Approximately -0.278 to Infinity | Modern networks |
| GELU | Approximately -0.17 to Infinity | Transformers |

---

# 18. Sigmoid Activation Function

## Formula

Sigmoid(x) = 1 / (1 + e^(-x))

## Properties

Output range = (0, 1)

Sigmoid(0) = 0.5

Large positive inputs produce outputs near 1.

Large negative inputs produce outputs near 0.

## Advantages

- Smooth and differentiable
- Useful for binary probability outputs
- Easy to interpret

## Disadvantages

- Can suffer from vanishing gradients
- Not zero-centered
- Can saturate for large input magnitudes

## Use Cases

- Fraud detection
- Customer churn
- Binary sentiment classification
- Binary medical prediction

---

# 19. Tanh Activation Function

## Formula

Tanh(x) = (e^x - e^(-x)) / (e^x + e^(-x))

## Properties

Output range = (-1, 1)

Tanh(0) = 0

Outputs are zero-centered.

## Advantages

- Supports positive and negative outputs
- Smooth and differentiable
- Common in recurrent network states

## Disadvantages

- Can suffer from vanishing gradients
- Saturates at large input magnitudes

---

# 20. ReLU Activation Function

ReLU stands for Rectified Linear Unit.

## Formula

ReLU(x) = max(0, x)

## Example

Input = -5

Output = 0

Input = 3

Output = 3

## Advantages

- Simple computation
- Efficient for many hidden layers
- Reduces saturation for positive inputs
- Commonly works well in deep networks

## Disadvantages

- Zero gradient for negative inputs
- Can experience dying ReLU
- Not differentiable exactly at zero

In practice, a subgradient convention is used at zero.

---

# 21. Leaky ReLU Activation Function

Leaky ReLU modifies ReLU to allow a small negative output.

## Formula

LeakyReLU(x) = x if x > 0
               alpha*x otherwise

Typical alpha = 0.01

## Example

Input = -5

Output = -0.05

Input = 5

Output = 5

## Advantage

Maintains a nonzero gradient for negative inputs.

This can reduce the dying ReLU problem.

---

# 22. Softmax Activation Function

Softmax converts a vector of class scores into
a probability distribution.

## Formula

Softmax(zi) = exp(zi) / Sum(exp(zj))

## Example

Model logits:

Cat = 2
Dog = 1
Bird = 0

Approximate probabilities:

Cat = 66.5%
Dog = 24.5%
Bird = 9.0%

Total = 100%

## Important

Softmax is appropriate when the output classes
are mutually exclusive.

For independent multiple labels, Sigmoid is
usually more appropriate.

---

# 23. What Is Forward Propagation?

Forward propagation is the process of passing input data
through the neural network to calculate predictions.

## Steps

1. Receive input features.
2. Multiply inputs by weights.
3. Add biases.
4. Apply activation functions.
5. Pass results to subsequent layers.
6. Generate the final prediction.
7. Calculate loss during training.

## Example

Inputs:

x1 = 2
x2 = 3

Weights:

w1 = 0.4
w2 = 0.6

Bias:

b = 0.5

Weighted sum:

z = (2 * 0.4) + (3 * 0.6) + 0.5

z = 0.8 + 1.8 + 0.5

z = 3.1

Apply Sigmoid:

Sigmoid(3.1) = 0.9569

Prediction = 0.9569

---

# 24. What Is a Loss Function?

A loss function measures how different a model prediction
is from the expected target.

The training process attempts to minimize loss.

## Example

Actual salary = 50000

Predicted salary = 45000

Prediction error = 5000

A loss function converts prediction differences
into a numerical training objective.

## Common Loss Functions

1. Mean Squared Error
2. Mean Absolute Error
3. Binary Cross-Entropy
4. Categorical Cross-Entropy
5. Sparse Categorical Cross-Entropy
6. Huber Loss
7. Hinge Loss
8. Kullback-Leibler Divergence

---

# 25. Mean Squared Error (MSE)

MSE is commonly used for regression.

## Formula

MSE = (1/n) * Sum((Actual - Predicted)^2)

## Example

Actual:

10, 20, 30

Predicted:

12, 18, 33

Squared errors:

(10 - 12)^2 = 4

(20 - 18)^2 = 4

(30 - 33)^2 = 9

MSE = (4 + 4 + 9) / 3

MSE = 5.6667

## Important

MSE penalizes larger errors more heavily because
errors are squared.

---

# 26. Mean Absolute Error (MAE)

MAE measures the average absolute prediction error.

## Formula

MAE = (1/n) * Sum(|Actual - Predicted|)

## Example

Actual:

10, 20, 30

Predicted:

12, 18, 33

Absolute errors:

2, 2, 3

MAE = (2 + 2 + 3) / 3

MAE = 2.
